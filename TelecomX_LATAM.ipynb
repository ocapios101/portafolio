```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
```

```python
df = pd.read_csv("TelecomX_LATAM.csv")  # Asegúrate de tener el archivo cargado en tu entorno
df.head()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="Churn", data=df)
plt.title("Distribución de Evasión de Clientes")
plt.xticks([0, 1], ["No Evasión", "Evasión"])
plt.xlabel("Churn")
plt.ylabel("Cantidad de Clientes")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

ruta_archivo = '/content/TelecomX_LATAM.csv'

try:
    # Intentar leer el CSV. Si tu CSV usa un separador diferente a la coma, especifícalo con 'sep='.
    # Puedes intentar diferentes separadores si no estás seguro (por ejemplo, sep=';' o sep='\t')
    df = pd.read_csv(ruta_archivo)
    print("Archivo CSV cargado exitosamente.")
    print("Columnas del DataFrame:")
    print(df.columns)
    print("\nPrimeras 10 filas:")
    print(df.head(10))

except FileNotFoundError:
    print(f"Error: No se encontró el archivo en la ruta especificada: {ruta_archivo}")
    print("Por favor, asegúrate de que el archivo 'TelecomX_LATAM.csv' esté en la carpeta '/content'.")
except Exception as e:
    print(f"Ocurrió un error al cargar el archivo CSV: {e}")
```

```python
fig, axs = plt.subplots(2, 2, figsize=(12, 8))

sns.boxplot(x="Churn", y="Soporte_tecnico", data=df, ax=axs[0, 0])
axs[0, 0].set_title("Soporte Técnico vs Churn")
axs[0, 0].set_xticklabels(["No", "Sí"])

sns.boxplot(x="Churn", y="Llamadas_reclamacion", data=df, ax=axs[0, 1])
axs[0, 1].set_title("Reclamaciones vs Churn")
axs[0, 1].set_xticklabels(["No", "Sí"])

sns.boxplot(x="Churn", y="Facturacion_mensual", data=df, ax=axs[1, 0])
axs[1, 0].set_title("Facturación vs Churn")
axs[1, 0].set_xticklabels(["No", "Sí"])

sns.boxplot(x="Churn", y="Promociones_recibidas", data=df, ax=axs[1, 1])
axs[1, 1].set_title("Promociones vs Churn")
axs[1, 1].set_xticklabels(["No", "Sí"])

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x="Internet", hue="Churn", data=df)
plt.title("Tipo de Internet vs Churn")
plt.xlabel("Internet")
plt.ylabel("Cantidad")
plt.legend(title="Churn", labels=["No", "Sí"])
plt.tight_layout()
plt.show()
```
